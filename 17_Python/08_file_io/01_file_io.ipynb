# 08. File Input / Output

* File 입출력은 AI 서버와 RAG 에서 굉장히 빈번하게 사용이 된다.
* AI Model 은 Data 가 많으면 많을 수록 정확도 있는 답변을 줄 수 있게 된다.
* 또한 회사에서는 Data 들이 다양한 형태 (.hwp, .csv, .txt, .pdf) 로 저장이 되어 있으며 
해당 파일들을 읽어 Model 을 학습시키게 된다.
* 공지사항 파일 읽기, 회사 내 파일 읽기


---

* RAG
* 회사 내규 파일 읽기 -> 문서 내용을 잘게 나누기 (Chunk)
* -> Embedding 생성 -> Vector DB 저장
* -> 사용자의 질문과 관련된 문서 Vector DB 에서 검색
* -> LLM 답변 생성

```python
# 파일 IO 관련 기본 문법
# 파일을 읽을 때는 open() 함수를 사용한다.
# with 
# java 에서의 try-with-resource 와 동일한 역할을 한다.
# 파일 사용이 끝나게 되면 자동으로 Stream (파일을 읽기 위한 통로 지원)
# 등 자원들을 종료시켜준다.

# open("파일명", "모드", "인코딩 방식")
with open("sample_question.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines) # 여러개의 내용이 반환될 때 리스트의 형태로 반환되는 것을 알 수 있다.

# public String test() {return "안녕"}
# test(); / 여기서는 String 이라고 한거니까 String 타입으로 나온다.
```

```python
# 주의사항!⚠️ : "w" 모도는 기존의 내용을 덮어쓴다.
answer = "내일의 날씨는 매우 습하고 폭우가 내리고있습니다"

with open("ai_answer.txt", "w", encoding="utf-8") as file:
    file.write(answer)

print("저장이 완료되었습니다")    
```

```python
# "a" 모드 : append 의 약자로서 기존 파일에 내용을 누적할 때 사용하게 된다.
answer = "내일 모래 날씨는 매우 어둡습니다."

with open("ai_answer.txt", "a", encoding="utf-8") as file:
    file.write(answer)
    # append 는 같은 행에 추가가 된다.
    # 만약 한 줄 띄우기를 하고자 하면 \n 개행 문자를 포함해야 한다.
print("저장이 완료되었습니다")
```

# 지금까지 배운 내용을 누적해서 적용!

```python
# dictionary 복습

with open("sample_question.txt", "r", encoding="utf-8") as file:
    question = file.read().strip() # 공백 삭제

# Map <String, Object > 역할을 하게 되며, Json 구조와 동일하다. {key-value}
response = {
    "question" : question,
    "answer" : f"{question} 에 대한 예시 답변"
}    

print(response)
```

```python
# List

# [] <- 자바에서는 배열이지만
# 파이썬에서는 List 이다.
documents = []

with open("rag_documents.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

# Python for 문은 Java 에서의 향샹된 for 문과 문법상 유사하다.    
# for in
for line in lines:
    documents.append(line.strip())

print(documents)
```

```python
class RAGService:
    def answer(self, question : str, documents : list) -> dict:

        context = "\n".join(documents)

        return {
            "question" : question,
            "context" : context,
            "answer" : "제공 된 문서를 바탕으로 LLM 이 답변을 생성했습니다."
        }
    
rag_service = RAGService()

# Spring Server 에 보내줄 AI 챗봇 결과 Json
response = rag_service.answer("RAG 가 도대체 무엇입니까?", documents)\

print(response)
```
