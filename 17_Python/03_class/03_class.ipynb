# 03. Class
* 관련 있는 함수 or 변수를 하나의 집합으로 묶는 문법이다.
* AI 관련 기능을 함수 여러 개로 흩어두지 않고
* 하나의 Service 객체로 묶어서 관리하기 위함

```python
class User:
    # __init__ : java 에서의 생성자 역할을 한다.
    # self : java 에서 this(자신의 래퍼런스 주소) 와 같은 역할을 한다.
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"안녕하세요. 제 이름은 {self.name} 입니다."


user01 = User("MNPPI223")


message = user01.introduce()

print(message)
```
