# 04. Router

* FastAPI 의 APIRouter 를 활용해서 APi 관련 코드를 분리하는 방법이다.
* 문제사항
    * main.py 파일에 모든 API 및 DTO 를 작성해두었다.
    * 프로젝트의 규모 및 기능이 추가 될 수록 main.py 가 복잡해진다.

* 필요사항
    * Route 의 개념이 필요하게 된다.
    * Spring Controller = FastAPI Router
    * APIRouter 를 사용해서 main.py 파일에 Router 를 등록한다.

* main.py
    * FastAPI 객체 생성
    * Router 등록    
* chat_router.py
    * ChatBot 관련 기능API
* rag_router.py
    * RAG 관련 기능 API

* __init__.py 파일을 각각의 폴더에 추가를 했다.
* Python 3.* 버전에서는 __init__.py 파일이 없게 된다면
* 단순한 폴더로만 인식을 하게 된다.
* __init__.py 파일이 있다면 단순 폴더가 아닌 Python Package 로 인식을 하기 때문에
* import 시 Error 가 발생하지 않는다.
