# 00_FastAPI 란?
* Python 기반의 API 서버를 만들 수 있는 웹 Framework
* Spring Framework 와 동일한 역할을 하게 된다.
---
* 빠르게 API 개발할 수 있다.
* 자동 Swagger 문서가 설정된다.
* Pydantic 기반의 요청/ 응답 검증 (Valid)
* 비동기 처리 지원
* Python 생태계이기 때문에 AI / ML / DL 관련 서버로 활용하기가 좋음. 즉, 라이브러리 / 래퍼런스가 풍부하다.

## 01_ 전체 흐름 구조
- -> : 요청이다
Client -> Next.js frontend -> Spring backend
-> FastAPI AI server -> PromptService 프롬포트 조립
-> ChatBotService 조립된 프롬포트와 함께 모델 호출
-> AI Model -> 답변 시작
